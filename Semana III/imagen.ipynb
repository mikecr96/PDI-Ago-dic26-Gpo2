```python
# Crear una imagen (100x100) negra y mostrarla

img2 = np.zeros((100, 100), dtype=np.uint8)
plt.imshow(img2, cmap='gray')
plt.show()
```

```python
# Crear una imagen (100x100) blanca y mostrarla

img3 = np.ones((100, 100), dtype=np.uint8) * 255
# print(img3)
plt.imshow(img3, cmap='gray', vmin=0)
plt.show()
```

```python
img4 = np.zeros((99, 99), dtype=np.uint8)
# slicing
img4[:33] = 255
img4[33:66] = 127
plt.imshow(img4.T, cmap='gray', vmin=0)
plt.show()
```

```python
img5 = np.zeros((99, 99), dtype=np.uint8)
# slicing
img5[:, :33] = 255
img5[:, 33:66] = 127
plt.imshow(img5, cmap='gray', vmin=0)
plt.show()
```
