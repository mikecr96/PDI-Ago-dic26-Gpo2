# Histogramas

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
tralalero = cv.imread('/Users/miguelcamargorojas/Downloads/tralalero.avif', 0)
```

```python
histograma = cv.calcHist([tralalero], [0], None, [256], (0, 256))
```
