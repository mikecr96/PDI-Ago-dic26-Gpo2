```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
tralalero = cv.imread('/Users/miguelcamargorojas/Downloads/tralalero.avif')
tralalero_gray = np.mean(tralalero, axis=2)
```

```python
plt.imshow(tralalero_gray, cmap='gray')
plt.show()
```

## Transformación negativa

s = L - 1 ｰ r

s = 256 - 1 - r

s = 255 - r

```python
filas, columnas = tralalero_gray.shape
tralalero_neg = tralalero_gray.copy()

for f in range(filas):
    for c in range(columnas):
        tralalero_neg[f, c] = 255 - tralalero_gray[f, c]

plt.imshow(np.hstack((tralalero_gray, tralalero_neg)), cmap='gray')
plt.show()
```

## Transformación log

s = c • log(1 +r)

c = 255/(log (1 + m))

```python
m = np.max(tralalero_gray)
C = 255 / np.log(1 + m)
C
```

```python
np.log(256)
```

```python
tralalero_log = tralalero_gray.copy()

for f in range(filas):
    for c in range(columnas):
        tralalero_log[f, c] = C * np.log(1 + tralalero_gray[f, c])
tralalero_log = tralalero_log.astype(np.uint8)
print(tralalero_log.dtype)
plt.imshow(tralalero_log, cmap='gray')
plt.show()
```
