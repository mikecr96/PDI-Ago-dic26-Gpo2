## Histogramas a color

```python
lena_bgr = cv.imread('/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo2/images/lena.jpg')
plt.imshow(lena_bgr[:, :, ::-1])
plt.show()
```

```python
histograma_b = cv.calcHist([lena_bgr], [0], None, [256], (0, 256))
plt.bar(range(256), histograma_b)
plt.title("Histograma del canal B")
plt.show()
```

```python
histograma_g = cv.calcHist([lena_bgr], [1], None, [256], (0, 256))
plt.bar(range(256), histograma_g)
plt.title("Histograma del canal G")
plt.show()
```

```python
histograma_r = cv.calcHist([lena_bgr], [2], None, [256], (0, 256))
plt.bar(range(256), histograma_r)
plt.title("Histograma del canal R")
plt.show()
```

```python
colores = ['b', 'g', 'r']

for contador, color in enumerate(colores):
    histograma = cv.calcHist([lena_bgr], [contador], None, [256], (0, 256))
    plt.bar(range(256), histograma, color=color)
    plt.title("Histograma de cada canal")
plt.show()
```
