```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
lena_gray = cv.imread('/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo2/images/lena.jpg', 0)
plt.imshow(lena_gray, cmap='gray')
plt.show()
```

```python
kernel = np.ones((3, 3), dtype=np.uint8) * 1/9
kernel.sum()
```

```python
lena_gray.size
```

```python
lena_filrada = cv.filter2D(lena_gray, -1, kernel)
```

```python
np.unique(lena_filrada)
```

```python
plt.imshow(np.hstack((lena_gray, lena_filrada)), cmap='gray')
plt.show()
```

```python
kernel5x5 = np.ones((5, 5)) * 1/25
lena_filrada = cv.filter2D(lena_gray, -1, kernel5x5)
plt.imshow(np.hstack((lena_gray, lena_filrada)), cmap='gray')
plt.show()
```

```python
def getKernel(n:int):
    return np.ones((n, n)) * 1 / (n*n)
```

```python
getKernel(9)
```

```python
lena_filrada = cv.filter2D(lena_gray, -1, getKernel(11))
plt.imshow(np.hstack((lena_gray, lena_filrada)), cmap='gray')
plt.show()
```
