## Laplaciano

```python
laplaciano1 = np.array([
    [0, -1, 0],
    [-1, 4, -1],
    [0, -1, 0]
])

laplaciano2 = laplaciano1 * -1

laplaciano3 = np.array([
    [-1, -1, -1],
    [-1, 8, -1],
    [-1, -1, -1]
])

laplaciano4 = laplaciano3 * -1

laplaciano1.sum()
```

```python
lena_l1 = cv.filter2D(lena, -1, laplaciano1)
```

```python
plt.imshow(lena_l1, cmap='gray')
plt.title('Lena filtro Laplaciano 1')
plt.show()
```

```python
np.unique(lena_l1)
```

```python
lena_l2 = cv.filter2D(lena, -1, laplaciano2)
plt.imshow(lena_l2, cmap='gray')
plt.title('Lena filtro Laplaciano 2')
plt.show()
```

```python
lena_l3 = cv.filter2D(lena, -1, laplaciano3)
plt.imshow(lena_l3, cmap='gray')
plt.title('Lena filtro Laplaciano 3')
plt.show()
```

```python
lena_l4 = cv.filter2D(lena, -1, laplaciano4)
plt.imshow(lena_l4, cmap='gray')
plt.title('Lena filtro Laplaciano 4')
plt.show()
```

```python
lena_l3 == lena_l4
```
