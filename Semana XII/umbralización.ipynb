```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Umbralización GLOBAL

```python
sudoku_gray = cv.imread('/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo2/images/sudoku.png', 0)
```

```python
def mostrar_img(img, title='Título genérico'):
    plt.imshow(img, 'gray')
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
mostrar_img(sudoku_gray)
```

```python
k = 50
sudoku_bin = cv.threshold(sudoku_gray, k, 255, cv.THRESH_BINARY)[1]
mostrar_img(sudoku_bin, f'Sudoku con umb {k}')
```

---

```python
lena_gray = cv.imread('/Users/miguelcamargorojas/iCloud Drive (Archive)/Documents/UP/PDI-Ago-dic26/PDI-Ago-dic26-Gpo2/images/lena.jpg', 0)
```

```python
k = int(lena_gray.mean())
lena_bin = cv.threshold(lena_gray, k, 255, cv.THRESH_BINARY)[1]
mostrar_img(lena_bin, f'Lena umbralizada con umbral={k}')
```

```python
umb, lena_bin = cv.threshold(lena_gray, 0, 255, cv.THRESH_BINARY | cv.THRESH_OTSU)
mostrar_img(lena_bin, f'Lena umbralizada con umbral de Otsu={int(umb)}')
```

```python
umb, lena_bin = cv.threshold(lena_gray, 0, 255, cv.THRESH_BINARY + cv.THRESH_TRIANGLE)
mostrar_img(lena_bin, f'Lena umbralizada con umbral de Triangle={int(umb)}')
```

```python
umb, sudoku_bin_otsu = cv.threshold(sudoku_gray, 0, 255, cv.THRESH_BINARY | cv.THRESH_OTSU)
mostrar_img(sudoku_bin_otsu, f'Sudoku umbralizado con umbral de Otsu={int(umb)}')
```

# Umbralización Adaptativa

```python
sudoku_gray_blur = cv.medianBlur(sudoku_gray, 5) # pre-procesamiento
sudoku_bin_7 = cv.adaptiveThreshold(sudoku_gray_blur, 255, cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, 11, 2)
mostrar_img(sudoku_bin_7, 'Sudoku binarizado adaptativamente con tamaño de bloque 11')
```
