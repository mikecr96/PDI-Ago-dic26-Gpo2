## Aislamiento cromático

```python
# Tralalero se vea roja
solo_r = tralalero.copy() # Para no modificar al original
# Apagar el canal GB, solo me quedo con el R
solo_r[:, :, 1] = 0
solo_r[:, :, 2] = 0
```

```python
solo_r.shape
```

```python
solo_r
```

```python
plt.imshow(solo_r, cmap='viridis')
plt.title('Tralalero solo canal rojo (R)')
plt.show()
```

```python
# Tralalero se vea azul
solo_b = tralalero.copy() # Para no modificar al original
# Apagar el canal RG, solo me quedo con el B
solo_b[:, :, [0, 1]] = 0
plt.imshow(solo_b)
plt.title('Tralalero solo canal azul (B)')
plt.show()
```

```python
# Tralalero se vea verde
solo_g = tralalero.copy() # Para no modificar al original
# Apagar el canal RB, solo me quedo con el G
solo_g[:, :, [0, 2]] = 0
plt.imshow(solo_g)
plt.title('Tralalero solo canal verde (G)')
plt.show()
```

## Mezcla de canales (Colores secundarios)

```python
# R + G (Sin azul) -> Amarillo

# Tralalero se vea amarillo
amarillo = tralalero.copy() # Para no modificar al original
# Apagar el canal B, solo me quedo con el RG
amarillo[:, :, 2] = 0
plt.imshow(amarillo)
plt.title('Tralalero amarillo (R + G)')
plt.show()
```

```python
# R + B -> Magenta
# G + B -> Cian
```
